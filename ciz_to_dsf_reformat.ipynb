### format new CIZ data to legacy DSF format 

```python
import pandas as pd
```

```python
# this only covers columns from old crsp.dsf.

CIZ_TO_DSF = {
    "dlycaldt": "date",
    "dlyprc": "prc",
    "dlylow": "bidlo",
    "dlyhigh": "askhi",
    "dlyvol": "vol",
    "dlyret": "ret",
    "dlyretx": "retx",
    "dlybid": "bid",
    "dlyask": "ask",
    "dlycumfacpr": "cfacpr",
    "dlycumfacshr": "cfacshr",
    "dlyopen": "openprc",
    "dlynumtrd": "numtrd",
    # permno, permco, shrout are unchanged
}
```

```python
list(CIZ_TO_DSF)
```

```python
path = "data/dsf/"
START_YEAR = 2010
END_YEAR = 2020
df_optimized = pd.read_parquet(
    path, 
    engine="pyarrow",
    columns=list(CIZ_TO_DSF),
    filters=[("year", ">=", START_YEAR), ("year", "<=", END_YEAR)]
)
df_optimized.head()
```

```python
# after data
df_optimized = df_optimized.rename(columns=CIZ_TO_DSF)
df_optimized.head()
```
